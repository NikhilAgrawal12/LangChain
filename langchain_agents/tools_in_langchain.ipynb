Custom Tools

```python
from langchain_core.tools import tool
```

```python
# step 1 - create a function

def multiply(a,b):
    """Multiply two numbers"""
    return a*b
```

```python
#Step 2 - add type hints

def multiply(a:int, b:int) -> int:
    """Multiply two numbers"""
    return a*b
```

```python
# Step 3 - add tool decorator

@tool
def multiply(a:int, b:int) -> int:
    """Multiply two numbers"""
    return a*b
```

```python
result = multiply.invoke({"a":3, "b":5})
```

```python
print(result)
```

```python
print(multiply.name)
print(multiply.description)
print(multiply.args)
```

```python
print(multiply.args_schema.model_json_schema())
```

Method 2 - Using Structured Tool

```python
from langchain_core.tools import StructuredTool
from pydantic import BaseModel, Field
```

```python
class MultiplyInput(BaseModel):
    a: int = Field(description="The first number to add")
    b: int = Field(description="The seccond number to add")
```

```python
#Not necessary to mention types in below tool function 
```

```python
def multiply_func(a:int, b:int) -> int:
    return a*b
```

```python
multiply_tool = StructuredTool.from_function(
    func=multiply_func,
    name="multiply",
    description="Multiply the numbers",
    args_schema=MultiplyInput
)
```

```python
result = multiply_tool.invoke({"a":3,"b":5})
```

```python
result
```

Method 3 - Using BaseTool Class

```python
from langchain.tools import BaseTool
from typing import Type
```

```python
class MultiplyInput(BaseModel):
    a: int = Field(description="The first number to add")
    b: int = Field(description="The second number to add")
```

```python
class MultiplyTool(BaseTool):
    name: str = "multiply"
    description: str = "Multiply two numbers"

    args_schema: Type[BaseModel] = MultiplyInput

    def _run(self,a:int,b:int) -> int:
        return a*b
```

```python
multiply_tool = MultiplyTool()
```

```python
result= multiply_tool.invoke({"a":5,"b":3})
```

```python
print(result)
```

Toolkit

```python
# Custom tools
@tool
def add(a:int,b:int)->int:
    """Add two numbers"""
    return a+b


@tool 
def mutiply(a:int,b:int)->int:
    """Multiply two numbers"""
    return a*b
```

```python
class MathToolkit:
    def get_tools(self):
        return [add,multiply]
```

```python
tools = MathToolkit()
```

```python
tools = tools.get_tools()
```

```python
for tool in tools:
    print(tool.name, "=>",tool.description)
```
